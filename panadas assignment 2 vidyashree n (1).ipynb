```python
df["country"] = df["country"].fillna(df["country"].mode()[0])
```

```python
df["agent"] = df["agent"].fillna(0)
```

```python
# 3. Handle missing company using 0
df["country"] = df["country"].fillna(df["country"].mode()[0])
df["agent"] = df["agent"].fillna(0)
df["company"] = df["company"].fillna(0)
df
```

```python
# 4. Convert agent and company to integer
df["agent"] = df["agent"].astype(int)
df["company"] = df["company"].astype(int)
df

```

```python
# 5. Text cleaning: strip spaces + title case
df["hotel"] = df["hotel"].str.strip().str.title()
df["meal"] = df["meal"].str.strip().str.upper()
df
```

```python
# 6. Remove duplicate rows
df.drop_duplicates(inplace=True)
df
```

```python
# 7. Fix invalid ADR values
df = df[df["adr"] >= 0]
df
```

```python
df.isna().sum()
```

```python
# Remove invalid ADR and create a clean copy
df = df[df["adr"] >= 0].copy()
df
```

```python
# Create total guests column
df["total_guests"] = df["adults"] + df["children"] + df["babies"]
df
```

```python
# 2. Booking status (conditional)
df["booking_status"] = np.where(df["is_canceled"] == 1, "Canceled", "Confirmed")
df
```

```python
#High value booking (conditional)
df["high_value_booking"] = df["adr"] > 150
df
```

```python
df["hotel"].value_counts()  #Insight: City Hotels receive more bookings than Resort Hotels.
```

```python
df.groupby("hotel")["is_canceled"].mean()  #Insight 2: Cancellation rate by hotel type (groupby + agg)
```

```python
df.groupby("hotel")["adr"].mean()  #Insight 3: Average ADR by hotel type (groupby)
```

```python
pd.pivot_table(
    df,
    index="hotel",
    columns="high_value_booking",
    values="adr",
    aggfunc="count",
    fill_value=0
)  #Insight: High-value bookings are more common in Resort Hotels.
```

```python
df[(df["total_guests"] >= 4) & (df["is_canceled"] == 0)].head()  #Insight: Large group bookings are more likely to be confirmed.
```
